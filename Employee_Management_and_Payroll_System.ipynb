```python
import mysql.connector

def add_employee():
    try:
        emp_id = int(input("Enter Employee ID: "))
        name = input("Enter Name: ")
        gender = input("Enter Gender: ")
        phone = input("Enter Phone: ")
        email = input("Enter Email: ")
        role = input("Enter Job Role: ")
        hire_date = input("Enter Hire Date (YYYY-MM-DD): ")
        salary = float(input("Enter Basic Salary: "))
        dept_id = int(input("Enter Department ID: "))

        query = """
        INSERT INTO employee
        VALUES (%s,%s,%s,%s,%s,%s,%s,%s,%s)
        """

        values = (emp_id, name, gender, phone,
                  email, role, hire_date,
                  salary, dept_id)

        cursor.execute(query, values)
        conn.commit()

        print("Employee Added Successfully!")

    except mysql.connector.Error as err:
        print("Database Error:", err)

    except ValueError:
        print("Please enter valid numeric values.")
```

```python
def search_employee():

    try:
        emp_id = int(input("Enter Employee ID: "))

        cursor.execute(
            "SELECT * FROM employee WHERE emp_id=%s",
            (emp_id,)
        )

        row = cursor.fetchone()

        if row:
            print(row)
        else:
            print("Employee Not Found!")

    except ValueError:
        print("Invalid Employee ID")
```

```python
def update_salary():

    try:
        emp_id = int(input("Employee ID: "))
        salary = float(input("New Salary: "))

        cursor.execute("""
        UPDATE employee
        SET basic_salary=%s
        WHERE emp_id=%s
        """, (salary, emp_id))

        if cursor.rowcount == 0:
            print("Employee Not Found!")
        else:
            conn.commit()
            print("Salary Updated!")

    except ValueError:
        print("Enter valid numbers.")
```

```python
def mark_attendance():

    attendance_id = int(input("Attendance ID: "))
    emp_id = int(input("Employee ID: "))
    date = input("Date (YYYY-MM-DD): ")
    status = input("Status (Present/Absent): ")

    query = """
    INSERT INTO attendance
    VALUES (%s,%s,%s,%s)
    """

    values = (attendance_id, emp_id, date, status)

    cursor.execute(query, values)
    conn.commit()

    print("Attendance Marked Successfully!")
```

```python
def view_attendance():

    query = """
    SELECT emp_id, attendance_date, status
    FROM attendance
    ORDER BY attendance_date
    """

    cursor.execute(query)

    rows = cursor.fetchall()

    print("\n------ ATTENDANCE ------")

    for row in rows:
        print(row)
```

```python
def attendance_report():

    query = """
    SELECT e.emp_name,
           a.attendance_date,
           a.status
    FROM employee e
    JOIN attendance a
    ON e.emp_id = a.emp_id
    ORDER BY a.attendance_date
    """

    cursor.execute(query)

    rows = cursor.fetchall()

    print("\n------ ATTENDANCE REPORT ------")

    for row in rows:
        print(row)
```

```python
def generate_payroll():

    payroll_id = int(input("Enter Payroll ID: "))
    emp_id = int(input("Enter Employee ID: "))
    month = input("Enter Month: ")

    # Get employee salary
    cursor.execute(
        "SELECT basic_salary FROM employee WHERE emp_id=%s",
        (emp_id,)
    )

    result = cursor.fetchone()

    if result is None:
        print("Employee Not Found!")
        return

    basic = float(result[0])

    hra = basic * 0.20
    da = basic * 0.10
    pf = basic * 0.12

    net_salary = basic + hra + da - pf

    query = """
    INSERT INTO payroll
    VALUES (%s,%s,%s,%s,%s,%s,%s,%s)
    """

    values = (
        payroll_id, emp_id, month,
        basic, hra, da, pf, net_salary
    )

    cursor.execute(query, values)
    conn.commit()

    print("\nPayroll Generated Successfully!")
    print(f"Net Salary : ₹{net_salary:.2f}")
```

```python
generate_payroll()
```

```python
def view_payroll():

    query = """
    SELECT e.emp_name,
           p.month,
           p.basic,
           p.hra,
           p.da,
           p.pf,
           p.net_salary
    FROM employee e
    JOIN payroll p
    ON e.emp_id = p.emp_id
    """

    cursor.execute(query)

    rows = cursor.fetchall()

    print("\n----------- PAYROLL REPORT -----------")

    for row in rows:
        print(row)
```

```python
view_payroll()
```

```python
def department_salary_report():

    query = """
    SELECT d.dept_name,
           ROUND(AVG(e.basic_salary),2)
    FROM employee e
    JOIN department d
    ON e.dept_id=d.dept_id
    GROUP BY d.dept_name
    """

    cursor.execute(query)

    rows = cursor.fetchall()

    print("\nAverage Salary by Department\n")

    for row in rows:
        print(row)
```

```python
def employee_count_report():

    query = """
    SELECT d.dept_name,
           COUNT(e.emp_id)
    FROM department d
    LEFT JOIN employee e
    ON d.dept_id = e.dept_id
    GROUP BY d.dept_name
    """

    cursor.execute(query)

    rows = cursor.fetchall()

    print("\nEmployee Count by Department\n")

    for row in rows:
        print(row)
```

```python
def highest_paid_employee():

    query = """
    SELECT emp_name,
           basic_salary
    FROM employee
    WHERE basic_salary = (
        SELECT MAX(basic_salary)
        FROM employee
    )
    """

    cursor.execute(query)

    row = cursor.fetchone()

    print("\nHighest Paid Employee\n")
    print(row)
```
